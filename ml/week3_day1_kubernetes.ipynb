```python
# ===================================================
# WEEK 3 DAY 1 - Kubernetes (K8s)
# Phetho Tlaka | May 2026
# 
# WHAT IS KUERBETS?
# Kubernetes (K8s) is a container orchestrator
# it manages multiple Docker containers across
# multiple machines automatically
# 
# WHAT IT SOLVES:
# Runs multiple copies of the container
# Auto-restarts crashed containers
# Balance traffic across containers
# Scales up when busy, down when quiet
# Zero downtime deployments 
# Self-healing infrastructure
# 
# KEY CONCEPTS 
# Pod          - smallest init, wraps 1+ containers
# Deployment   - manages pods, ensures N copies run
# Service      - exposes pods to the network
# Node         - a machine inside a cluster
# Cluster      - group of nodes Kubernetes manages
# ===============================================================

print("=" * 52)
print("  WEEK 3 DAY 1 - Kubernetes")
print("=" * 52)
print()
print("Key Kubernetes concepts:")
print()
concepts = {
    "Pod":        "Smallest unit - wraps 1+ containers",
    "Deployment": "Manages pods - ensures N copies run",
    "Service":    "Expose pods to the network",
    "Node":       "A machine in the cluster",
    "Cluster":    "Group of nodes K8s manage",
    "kubectl":    "Command line tool to contol K8s",
    "Namespace":  "Virtual Cluster within a cluster",
    "ReplicaSet": "Ensures N identical pod alwyas run",
}
for concept, desc in concepts.items():
    print(f"  {concept:12} → {desc}")
print()
print("Docker  = one chef cooking alone")
print("K8s     = manger hiring/firing chefs")
print("         automatically based on demand")
```

```python
# ============================================================
# STEP 1: INSTALL MINIKUBE 
# ===========================================================
# minikube = Kubernetes running on your single VM 
# Same as real production K8s - just machine 

import subprocess
import os

print("=" * 50)
print("  STEP 1: INSTALL MINIKUBE")
print("=" * 50)
print()

# Check if minikube is already installed
result = os.popen('minikube version').read().strip()

if 'minikube' in result:
    print(f"Already installed: {result}")
else: 
    print("Installing minikube...")
    print("This takes 1 minutes")
    print()

    #Download minikube binary 
    os.system(
       'curl -LO https://strorage.googleapi.com/'
       'minikube/releases/latest/'
       'minikube-linux-am64'
    )

   # Install it
    os.system(
       'sudo install minikube-linux-amd64 '
       '/usr/local/bin/minikube'
    )

   # Verify 
    result = os.popen('minikube version').read()
    print(result)
    print("Minikube installed!")

print()
print("Next: install kubectl")
print("kubectl = your tool to talk to Kubernetes")

# Check kubectl
kube = os.popen('kubectl version --client').read()
if 'Client' in kube:
    print("kubectl already installed1")
    print(kube)
else:
    print("Installing kubctl...")
    os.system(
        'curl -LO "https://d1.k8s.io/release/'
        '$(curl -L -s https://dl.k8s.io/release/'
        'stable.txt)/bin/linux/amd64/kubectl"'
    )
    os.system('sudo install kubectl /usr/local/bin/')
    print("kubectl install!")
```
